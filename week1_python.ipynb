```python
import pandas as pd
url = "https://raw.githubusercontent.com/ImMatt-blip/ba-ai-project/refs/heads/main/AI_SocialMedia_Student_Dataset%5B1%5D.csv"
df=pd.read_csv(url)
df.head()    # first 5 rows
```

```python
df.info()  # columns, types, blanks
```

```python
print(df.duplicated().sum(),  "duplicates")
print(df.isna().sum())  # blanks per column

df=df.drop_duplicates()
df=df.dropna()
print(df.shape)    # rows, columns left
```

```python
df.groupby("Education_Level")["Mental_Health_Score"].sum().plot(
    kind="bar",title="Total Mental Health Score By Education Level")
```
